```python
from sklearn.preprocessing import LabelEncoder 
lbl_encoder = LabelEncoder() 
```

```python
import pandas as pd 
df = pd.DataFrame({
    'colors': ['red','blue','green','green','blue', 'green'] 
})
```

```python
lbl_encoder.fit_transform(df[['colors']])
```

```python
lbl_encoder.transform([['red']])
```

## Ordinal Encoding 
Eg: 
1. High school:1
2. Collage: 2 
3. Graduate: 3 
4. Posr-graduate-4


```python
## Ordinal Encoding
from sklearn.preprocessing import OrdinalEncoder 
```

```python
df = pd.DataFrame({
    'size': ['small', 'medium', 'large', 'medium', 'small', 'large']
})
```

```python
df
```

```python
## Craete an instance of Ordinal Encoding and then fit_tranform  
encoder = OrdinalEncoder(categories=[['small', 'medium', 'large']])
```

```python
encoder.fit_transform(df[['size']])
```
